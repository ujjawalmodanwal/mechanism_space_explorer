```python
# Execution settings
BUNDLE_DIR = "/kaggle/input/<dataset>"
ENGINE = "mace"
SYSTEMS = ["hcooh", "hcooh_h2o", "hcho_nh3"]
RESUME_ZIP = None
# A short first run checks installs and GPU use, e.g. 0.3 with ["hcooh"].
DEADLINE_HOURS = 11.5
```

```python
import glob
import subprocess
import sys

# Kaggle may unpack the uploaded zip one folder down.
pattern = f"{BUNDLE_DIR}/**/scripts/kaggle/run_refine.py"
runner = glob.glob(pattern, recursive=True)
msg = f"run_refine.py not found once under {BUNDLE_DIR}: {runner}"
assert len(runner) == 1, msg
cmd = [
    sys.executable,
    runner[0],
    "--bundle-dir",
    BUNDLE_DIR,
    "--engine",
    ENGINE,
    "--systems",
    *SYSTEMS,
    "--deadline-hours",
    str(DEADLINE_HOURS),
]
if RESUME_ZIP:
    cmd.extend(["--resume-zip", RESUME_ZIP])

print("Running:", " ".join(cmd))
res = subprocess.run(cmd)
if res.returncode != 0:
    print(f"WARNING: runner failed with exit code {res.returncode}")
```
